```python
import os
import pandas as pd

dir_path = '/storage/lingyuan2/STATES_analysis/AD_downstream/12_SDEG'

filtered_genes = []

for fname in os.listdir(dir_path):
    if fname.endswith('30ring_14mAD.csv'):
        fpath = os.path.join(dir_path, fname)
        try:
            df = pd.read_csv(fpath)
        except Exception:
            continue
        if 'gene' not in df.columns:
            if 'Gene' in df.columns:
                df['gene'] = df['Gene']
        if all(col in df.columns for col in ['gene','padj','log2FC']):
            sel = df[(df['padj'] < 0.01) & (df['log2FC'] > 0.5)].copy()
            if not sel.empty:
                sel['source_file'] = fname
                filtered_genes.append(sel)

if filtered_genes:
    result_df = pd.concat(filtered_genes, ignore_index=True)
else:
    result_df = pd.DataFrame()
```

```python
result_df.to_csv("ring30_14mAD_SDEG_totalRNA.csv", index=False)
```
